```python
import random

print("Hand Cricket Game")
print("Enter number between 0-6")

class Player:
    def __init__(self, name):
        self.name = name
        self._score = 0

    def add_runs(self, runs):
        self._score += runs

    def score(self):
        return self._score

class P1(Player):
    def play(self):
        while True:
            try:
                num = int(input(f"{self.name}, enter your number (0-6): "))
                if 0 <= num <= 6:
                    return num
                else:
                    print("Invalid input, try again.")
            except ValueError:
                print("Invalid input, enter a number between 0 and 6.")


class P2(Player):
    def play(self):
        while True:
            try:
                num = int(input(f"{self.name}, enter your number (0-6): "))
                if 0 <= num <= 6:
                    return num
                else:
                    print("Invalid input, try again.")
            except ValueError:
                print("Invalid input, enter a number between 0 and 6.")


def turns(batsman: Player, bowler: Player, target=None):
    bat = batsman.play()
    bowl = bowler.play()

    if bat == bowl:
        print(f"\n{batsman.name} is OUT! Final score: {batsman.score()}\n")
        return False
    else:
        batsman.add_runs(bat)
        print(f"{batsman.name} hits {bat} → Total: {batsman.score()}")
        if target and batsman.score() >= target:
            print(f"\n{batsman.name} chased the target!\n")
            return False
        return True


class Handcricket:
    def __init__(self):
        self.player1 = P1("Player 1")
        self.player2 = P2("Player 2")

    def start(self):
        print("\n--- Welcome to Hand Cricket ---\n")
        while turns(self.player1, self.player2):
            pass
        target = self.player1.score() + 1
        print(f"Target for {self.player2.name}: {target}\n")
        while turns(self.player2, self.player1, target):
            pass
        print("\n--- Final Scores ---")
        print(f"{self.player1.name}: {self.player1.score()}")
        print(f"{self.player2.name}: {self.player2.score()}")
        if self.player1.score() > self.player2.score():
            print("Player 1 wins!")
        elif self.player2.score() > self.player1.score():
            print("Player 2 wins!")
        else:
            print("it's a tie!")


game = Handcricket()
game.start()
```
